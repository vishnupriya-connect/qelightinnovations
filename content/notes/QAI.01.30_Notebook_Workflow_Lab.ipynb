Goal: clean three course questions, classify them, check normal and boundary cases, and produce a result another learner can reproduce. Restart the kernel and run all cells in order.


```python
raw_questions = ["  Explain   GenAI  ", "   ", "What is RAG?"]
max_characters = 80
print("Questions received:", len(raw_questions))
```

## Functions

Trim surrounding whitespace and collapse repeated internal whitespace before applying the maximum length rule.


```python
def clean_question(raw):
    return " ".join(raw.split())

def classify_question(raw, max_characters):
    cleaned = clean_question(raw)
    if not cleaned:
        return "empty", cleaned
    if len(cleaned) > max_characters:
        return "too_long", cleaned
    return "ready", cleaned
```

## Process the inputs


```python
classified = [classify_question(raw, max_characters) for raw in raw_questions]
ready_questions = [cleaned for status, cleaned in classified if status == "ready"]
```

## Check five claims and display the result

For the execution-order exercise, restart the kernel and run this last code cell first. Observe NameError. Restart again and run all cells in page order.


```python
assert classify_question("  Explain   GenAI  ", max_characters) == ("ready", "Explain GenAI")
assert classify_question("   ", max_characters) == ("empty", "")
assert classify_question("x" * 80, max_characters)[0] == "ready"
assert classify_question("x" * 81, max_characters)[0] == "too_long"
assert raw_questions[0] == "  Explain   GenAI  "
print("Checks passed: 5")
print("Ready questions:", ready_questions)
```
